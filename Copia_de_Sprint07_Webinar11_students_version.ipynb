````python
# Automatización con Funcion
col_num=['unidades', 'precio_unitario', 'descuento_pct', 'ingreso', 'costo', 'margen', 'edad_cliente', 'satisfaccion', 'dias_entrega']

def calcular_mean_std_min_max(df_limpio, col_num):
    resultados = {}

    for col in col_num:
        resultados[col] = {
            'mean': df_limpio[col].mean(),
            'std': df_limpio[col].std(),
            'max': df_limpio[col].max(),
            'min': df_limpio[col].min()
        }

    return resultados
````

````python
resultado = calcular_mean_std_min_max(df_limpio, col_num)

print(resultado)
````

## Ejercicio 8 · Outliers con IQR \(8 min\)

La regla IQR permite detectar valores inusualmente altos o bajos.

No significa que esos valores estén mal. Significa que deben revisarse antes de sacar conclusiones.

### Espacio de trabajo del estudiante

**Mi interpretación de la consigna:**  

**Mi procedimiento / fórmula / consulta:**  

**Resultado obtenido:**  

**Conclusión breve:**  


````python
# TODO estudiante: escribe aquí tu código.
# Función para calcular límites IQR

def calcular_limites_iqr(tabla, columna):
    q1 = tabla[columna].quantile(0.25)
    q3 = tabla[columna].quantile(0.75)

    IQR = q3 - q1

    limite_inferior = q1-1.5*IQR
    limite_superior = q3+1.5*IQR

    return limite_inferior, limite_superior
````

````python
down_limit, up_limit = calcular_limites_iqr(df_limpio, 'ingreso')


print('Limite Inferior:', down_limit)
print('Limite Superior:', up_limit)
````

````python
# Indentificar Ventas Atípicas por Ingreso
outliers = df_limpio[
    (df_limpio['ingreso'] < down_limit) |
    (df_limpio['ingreso'] > up_limit)
    ]

print('Cantidad de posibles outliers:', len(outliers))
outliers

````

## Ejercicio 9 · Segmentación de ventas \(8 min\)

Segmentar ayuda a convertir números en etiquetas interpretables. Vamos a clasificar cada venta por tamaño de ingreso.

### Espacio de trabajo del estudiante

**Mi interpretación de la consigna:**  

**Mi procedimiento / fórmula / consulta:**  

**Resultado obtenido:**  

**Conclusión breve:**  


````python
# Función para segmentación
def clasificar_venta(ingreso):
    if ingreso >= 4_000_000:
        return 'Alta'
    elif ingreso >= 1_500_000:
        return 'Media'
    else:
        return 'Baja'


# Aplicar funcion a cada venta
df_limpio['segmento_venta'] = df_limpio['ingreso'].apply(clasificar_venta)

df_limpio[['venta_id','ingreso','segmento_venta']]
````

````python
# Conteo por Segmento

conteo_segmento = df_limpio['segmento_venta'].value_counts()

display(conteo_segmento)

conteo_segmento.plot(
    kind='bar',
    title='Cantidad de ventas por segmento',
    figsize=(8,4)
)
plt.xlabel('Segmento')
plt.ylabel('Cantidad Ventas')
plt.show()
````
